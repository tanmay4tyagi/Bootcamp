```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Load sample image
image_path = 'sample_input.png'
if not os.path.exists(image_path):
    image_path = r'C:\Users\dd\Desktop\Gemini.png'

img = cv2.imread(image_path)
height, width, channels = img.shape

print(f"Loaded Image: {os.path.basename(image_path)}")
print(f"Dimensions  : {width} x {height} pixels | Channels: {channels}")

# Convert to RGB for display in Matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(6, 4))
plt.imshow(img_rgb)
plt.title("Original Image", fontsize=12)
plt.axis("off")
plt.show()
```

---
### 1. Geometric Transformations (Resizing, Cropping, Flipping)
- **`cv2.resize`**: Scale image dimensions using bilinear interpolation.
- **ROI Slicing**: Crop central region using NumPy indexing.
- **`cv2.flip`**: Reflect across horizontal (`1`), vertical (`0`), or both (`-1`) axes.


```python
# Resizing
resized_img = cv2.resize(img, (600, 350), interpolation=cv2.INTER_LINEAR)

# Central Cropping (ROI)
y1, y2 = int(height * 0.15), int(height * 0.85)
x1, x2 = int(width * 0.15), int(width * 0.85)
cropped_img = img[y1:y2, x1:x2]

# Horizontal Flipping
flipped_img = cv2.flip(img, 1)

# Display Comparison
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(cv2.cvtColor(resized_img, cv2.COLOR_BGR2RGB))
axes[0].set_title("Resized (600x350)")
axes[0].axis("off")

axes[1].imshow(cv2.cvtColor(cropped_img, cv2.COLOR_BGR2RGB))
axes[1].set_title("Central Cropped (ROI)")
axes[1].axis("off")

axes[2].imshow(cv2.cvtColor(flipped_img, cv2.COLOR_BGR2RGB))
axes[2].set_title("Flipped (Horizontal)")
axes[2].axis("off")

plt.tight_layout()
plt.show()
```

---
### 2. Color Spaces & Tonal Adjustments
- **Grayscale (`cv2.cvtColor`)**: Discard chromatic channels, retaining luminance.
- **Color Inversion (`cv2.bitwise_not`)**: Creates photographic negative.
- **Brightness & Contrast (`cv2.convertScaleAbs`)**: Adjust dynamic range.
- **Sepia Tone**: Multiply by standard vintage color matrix.


```python
# Grayscale
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Photographic Inversion (Negative)
inverted_img = cv2.bitwise_not(img)

# Contrast & Brightness Enhancement
contrast_img = cv2.convertScaleAbs(img, alpha=1.35, beta=35)

# Sepia Vintage Filter
sepia_kernel = np.array([
    [0.272, 0.534, 0.131],
    [0.349, 0.686, 0.168],
    [0.393, 0.769, 0.189]
])
sepia_img = np.clip(cv2.transform(img, sepia_kernel), 0, 255).astype(np.uint8)

fig, axes = plt.subplots(1, 4, figsize=(18, 4))
axes[0].imshow(gray_img, cmap='gray')
axes[0].set_title("Grayscale")
axes[0].axis("off")

axes[1].imshow(cv2.cvtColor(inverted_img, cv2.COLOR_BGR2RGB))
axes[1].set_title("Color Inversion (Negative)")
axes[1].axis("off")

axes[2].imshow(cv2.cvtColor(contrast_img, cv2.COLOR_BGR2RGB))
axes[2].set_title("Enhanced Contrast & Brightness")
axes[2].axis("off")

axes[3].imshow(cv2.cvtColor(sepia_img, cv2.COLOR_BGR2RGB))
axes[3].set_title("Vintage Sepia Tone")
axes[3].axis("off")

plt.tight_layout()
plt.show()
```

---
### 3. Filtering, Smoothing, Edge Detection & Sharpening
- **Gaussian Blur (`cv2.GaussianBlur`)**: Low-pass filter for noise suppression.
- **Bilateral Filter (`cv2.bilateralFilter`)**: Edge-preserving smoothing filter.
- **Sharpening (`cv2.filter2D`)**: High-pass kernel boosting edge gradients.
- **Canny Edge Detection (`cv2.Canny`)**: Multi-stage edge extraction.


```python
# Gaussian Blur
gaussian_blur = cv2.GaussianBlur(img, (15, 15), 0)

# Bilateral Edge-Preserving Smoothing
bilateral_smooth = cv2.bilateralFilter(img, d=9, sigmaColor=75, sigmaSpace=75)

# Sharpening
sharpen_kernel = np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]])
sharpened_img = cv2.filter2D(img, -1, sharpen_kernel)

# Canny Edge Detection
canny_edges = cv2.Canny(gray_img, 100, 200)

fig, axes = plt.subplots(1, 4, figsize=(18, 4))
axes[0].imshow(cv2.cvtColor(gaussian_blur, cv2.COLOR_BGR2RGB))
axes[0].set_title("Gaussian Blur")
axes[0].axis("off")

axes[1].imshow(cv2.cvtColor(bilateral_smooth, cv2.COLOR_BGR2RGB))
axes[1].set_title("Bilateral (Edge-Preserving)")
axes[1].axis("off")

axes[2].imshow(cv2.cvtColor(sharpened_img, cv2.COLOR_BGR2RGB))
axes[2].set_title("Detail Sharpening")
axes[2].axis("off")

axes[3].imshow(canny_edges, cmap='gray')
axes[3].set_title("Canny Edge Detection")
axes[3].axis("off")

plt.tight_layout()
plt.show()
```
