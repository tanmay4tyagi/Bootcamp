### 3. Use pandas features for data cleaning (Hint-RegEx)

```python
import pandas as pd
# Load the dataset to clean
df_clean = pd.read_csv('khubh_mela_data.csv')

# 1. Clean 'Name': remove special characters, dots, and slashes using RegEx
df_clean['Name'] = df_clean['Name'].str.replace(r'[^a-zA-Z\s]', '', regex=True)

# 2. Replace multiple spaces with a single space and strip whitespace
df_clean['Name'] = df_clean['Name'].str.replace(r'\s+', ' ', regex=True).str.strip().str.title()

# 3. Standardize 'City' names using RegEx (e.g. Chenai/Chennaai -> Chennai, NDL/New Delhi -> Delhi, Bombay -> Mumbai)
df_clean['City'] = df_clean['City'].str.replace(r'Chen+a+i', 'Chennai', regex=True)
df_clean['City'] = df_clean['City'].str.replace(r'^(NDL|New Delhi)$', 'Delhi', regex=True)
df_clean['City'] = df_clean['City'].str.replace(r'^Bombay$', 'Mumbai', regex=True)

# 4. Standardize 'State' spelling using RegEx (e.g. Maharastra -> Maharashtra)
df_clean['State'] = df_clean['State'].str.replace(r'Maharastra', 'Maharashtra', regex=True)

print("Cleaned Data Preview:")
df_clean[['Visitor_ID', 'Name', 'City', 'State']].head(10)
```

### 4. Use Matplotlib and draw a Sine wave

```python
import matplotlib.pyplot as plt

# Create 500 equally spaced time values from 0 to 10
time = np.linspace(0, 10, 500)

# Calculate the amplitude of the sine wave
amplitude = np.sin(2 * np.pi * time)

# Plot the sine wave
plt.plot(time, amplitude)

# Add labels and title
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.title("Sine Wave")

# Display the graph
plt.show()
```
